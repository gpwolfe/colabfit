```python
PROPS = [
    "dimer_charge",
    "dimer_multiplicity",
    "momomer_a_charge",
    "monomer_a_multiplicity",
    "momomer_b_charge",
    "monomer_b_multiplicity",
    "natoms_monomer_a",
    "natoms_monomer_b",
    "CCSD(T)/CBS",
    "CCSD(T)/haTZ",  # heavy-aug-cc-pVTZ
    "MP2/haTZ",  # heavy-aug-cc-pVTZ
    "MP2/CBS",
    "MP2/aTZ",  #  aug-cc-pVTZ
    "MP2/aQZ",  #  aug-cc-pVQZ
    "HF/haTZ",  # heavy-aug-cc-pVTZ
    "HF/aTZ",  #  aug-cc-pVTZ
    "HF/aQZ",  #  aug-cc-pVQZ
    "sapt_total",  #  aug-cc-pVDZ
    "sapt_electrostatics",
    "sapt_exchange",
    "sapt_induction",
    "sapt_dispersion",
]


def nenci_props_parser(string):
    s = [float(x) for x in string.split()]
    return dict(zip(PROPS, s))


def reader(file_path):
    print(file_path.name)
    with open(file_path, "r") as f:
        atom = next(read_xyz(f, properties_parser=nenci_props_parser))
    atom.info["name"] = file_path.stem
    yield atom
```

```python
lens = []
for fp in Path("data/nenci2021/xyzfiles").glob("*.xyz"):
    lens.append((fp.stem, len(next(reader(fp)))))
```

```python
with open("lens.txt", "w") as f:
    f.writelines([x + "\n" for x in lens])
```
